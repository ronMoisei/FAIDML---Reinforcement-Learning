```python
import gymnasium as gym


def inspect_hopper():
    # Initialize the standard Hopper environment
    env = gym.make('Hopper-v4')
    env.reset()

    print("--- 1. State Space ---")
    obs_space = env.observation_space
    print(f"Type: {type(obs_space).__name__}")
    print(f"Shape: {obs_space.shape}")
    print(f"Continuous? {isinstance(obs_space, gym.spaces.Box)}")

    print("\n--- 2. Action Space ---")
    act_space = env.action_space
    print(f"Type: {type(act_space).__name__}")
    print(f"Shape: {act_space.shape}")
    print(f"Continuous? {isinstance(act_space, gym.spaces.Box)}")

    print("\n--- 3. Link Masses ---")
    # Note: In modern Gymnasium, we access the model directly via `env.unwrapped.model`
    # instead of the deprecated `env.sim.model`.
    model = env.unwrapped.model

    # Extract body names and their corresponding masses
    body_names = [model.body(i).name for i in range(model.nbody)]
    body_masses = model.body_mass

    for name, mass in zip(body_names, body_masses):
        print(f"Body: {name:10} | Mass: {mass:.4f} kg")

    # Run a quick random policy to ensure physics are stepping correctly
    print("\n--- Running Random Policy ---")
    for _ in range(5):
        action = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(action)
        print(f"Action: {action} | Reward: {reward:.4f}")
        if terminated or truncated:
            env.reset()

    env.close()


inspect_hopper()
```

*What is the state space in the Hopper environment? Is it discrete or continuous?*
It is Continuous (represented by a Box space). The standard observation is an 11-dimensional vector containing positional values (z-coordinate, joint angles) and velocities (angular velocities of the joints).

*What is the action space in the Hopper environment? Is it discrete or continuous?*
It is Continuous (represented by a Box space). The action is a 3-dimensional vector representing the continuous torques applied to the three motorized joints (thigh, leg, and foot).

*What is the mass value of each link of the Hopper environment?*

* world / torso: ~3.53 kg
* thigh: ~3.93 kg
* leg: ~2.71 kg
* foot: ~5.09 kg
